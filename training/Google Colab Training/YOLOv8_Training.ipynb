```python
DRIVE_PROJECT_ROOT = "/content/drive/MyDrive/dissertation_project"
DATASET_DIR = os.path.join(DRIVE_PROJECT_ROOT, "BTXRD_YOLO_MultiClass")
DATA_YAML_PATH = os.path.join(DATASET_DIR, "data.yaml")
```

```python
MODEL_TYPE = 'yolov8s-seg.pt'
IMAGE_SIZE = 640
BATCH_SIZE = 16
EPOCHS = 100
PATIENCE_EPOCHS = 30
OPTIMIZER = 'AdamW'
LEARNING_RATE = 0.001
PROJECT_NAME = "Bone_Tumor_MultiClass_YOLO"
EXPERIMENT_NAME = "YOLOv8"
DEGREES_ROTATION = 15.0
TRANSLATE_IMG = 0.1
SCALE_IMG = 0.5
FLIP_LR = 0.5
FLIP_UD = 0.5
HSV_H = 0.015
HSV_S = 0.7
HSV_V = 0.4
```

```python
model = YOLO(MODEL_TYPE)
results = model.train( data=DATA_YAML_PATH, epochs=EPOCHS, imgsz=IMAGE_SIZE, batch=BATCH_SIZE, project=PROJECT_NAME, name=EXPERIMENT_NAME, exist_ok=True, optimizer=OPTIMIZER, lr0=LEARNING_RATE, patience=PATIENCE_EPOCHS, degrees=DEGREES_ROTATION, translate=TRANSLATE_IMG, scale=SCALE_IMG, fliplr=FLIP_LR, flipud=FLIP_UD, hsv_h=HSV_H, hsv_s=HSV_S, hsv_v=HSV_V)
```
